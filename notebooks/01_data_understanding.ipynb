# Problems Found in the Raw Dataset

Based on the initial inspection, the following data-quality issues
were identified:

1. The dataset contains 94,948 records and 47 columns, which differs
   from the expected 89,625 records and 46 columns in the project plan.

2. Two records have missing game names.

3. There are 756 duplicate game names, while appid has no duplicates.

4. Several columns have substantial missing values:
   score_rank, metacritic_url, reviews, notes, website and support_url.

5. The description fields have moderate missing values:
   short_description, about_the_game and detailed_description.

6. genres and categories are stored as string representations of lists.

7. tags is stored as a string representation of a dictionary containing
   tag vote counts.

8. estimated_owners is stored as a string range rather than a numeric value.

9. release_date is stored as an object/string and will need conversion
   to datetime.

10. Price values are decimal monetary values and need to be interpreted
    consistently before INR conversion.

11. Some numeric fields may contain placeholder values such as 0 that
    need to be investigated rather than treated automatically as valid
    measurements.
